```python
#to print colors in set 1 but not in set 2
color_set1={"red","blue","green","yellow","black"}
color_set2={"blue","yellow","white"}
print("colors in set 1 but not in set 2:",color_set1-color_set2)
```

```python
#to print all colors from color_set1 and color_set 2
color_set1={"red","blue","green","yellow","black"}
color_set2={"blue","yellow","white","green"}
print("colors in both sets:",color_set1|color_set2)
```

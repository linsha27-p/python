```python
#count the occurrences of each word in a line of text
text = input("Enter a line of text: ")
words = text.split()
count = {}
for word in words:
    if word in count:
        count[word] = count[word] + 1
    else:
        count[word] = 1
for word in count:
    print(word, ":", count[word])
```

```python
#replace values greater than 100 with 'over'
numbers = list(map(int, input("Enter numbers: ").split()))
for i in range(len(numbers)):
    if numbers[i] > 100:
        numbers[i] = "over"
print(numbers)
```

```python
#count occurrences of 'a' in a list of first names
names = input("Enter first names: ").split()
count = 0
for name in names:
    for ch in name:
        if ch == 'a':
            count = count + 1
print("Number of 'a' occurrences:", count)
```

```python
#find GCD of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
gcd = 1
for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        gcd = i
print("GCD =", gcd)
```

```python
#create a list removing even numbers
numbers = list(map(int, input("Enter numbers: ").split()))
new_list = []
for num in numbers:
    if num % 2 != 0:
        new_list.append(num)
print("List after removing even numbers:", new_list)
```

```python
#find tha factorial of a number
n = int(input("Enter a number: "))
fact = 1
for i in range(1, n + 1):
    fact = fact * i
print("Factorial =", fact)
```

```python
#generate fibonacci series of N terms
n = int(input("Enter number of terms: "))
a = 0
b = 1
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
#find the sum of all items in a list
numbers = list(map(int, input("Enter numbers: ").split()))
sum = 0
for num in numbers:
    sum = sum + num
print("Sum =", sum)
```

```python
#four-digit numbers with all digits even and perfect square
import math
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))
print("starting range",start,"\nending range:",end)
print("result is:",end=" ")
for n in range(start,end+1):
    if int(math.sqrt(n))**2==n:
        if all(d in "02468" for d in str(n)):
            print(n,end=" ")
```

```python
#compare two list
list1 = list(map(int, input("Enter first list: ").split()))
list2 = list(map(int, input("Enter second list: ").split()))
if len(list1) != len(list2):
    print("Lists are not equal")
else:
    same = True
    for i in range(len(list1)):
        if list1[i] != list2[i]:
            same = False
            break
    if same:
        print("Lists are equal")
    else:
        print("Lists are not equal")
```

```python
#accept 5 students name and roll numbers
roll=input("Enter roll-no(Space separated):").split()
name=input("Enter names(Space separated):").split()
mydict=dict(zip(roll,name))
print("ascending order:",dict(sorted(mydict.items())))
print("descending order:",dict(sorted(mydict.items(),reverse=True)))
```

```python
#merge two dictinaries using update()
dict1={1:"lin",2:"sumu",3:"mika"}
dict2={4:"riya",5:"bis"}
print("First dictionary:",dict1)
print("Second dictionary:",dict2)
dict1.update(dict2)
print("Merged dictionaries:",dict1)
```

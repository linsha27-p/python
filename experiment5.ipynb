```python
#get 3rd element from first and 3rd element from last of a tuple
tuples=(1,2,3,4,5,6)
print("3rd element from the first is",tuples[2],"and 3rd element from the last is",tuples[-3])
```

```python
#generate a tuple and find its average
number=input("enter the number:")
tuples=tuple(map(int,number.split(",")))
print("tuple is:",tuples)
print("average is:",sum(tuples)/len(tuples))
```

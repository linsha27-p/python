```python
#first and last interchange
s=input("enter a string:")
print("new string is:",s[-1]+s[1:-1]+s[0])
```

```python
#crete a string from first 2 and last 2 charcter of a string
s=input("enter  string:")
print("new string is:",s[:2]+s[-2:])
```

```python
#create a single string seperated with space from two strings by swapping
s1=input("enter the first string:")
s2=input("enter the second string:")
print("new string is:",s2[0]+s1[1:]," ",s1[0]+s2[1:])
```

```python
#replace repeated first character with $
s=input("enter a string:")
print("new string is:",s[0]+s[1:].replace(s[0],'$'))
```
